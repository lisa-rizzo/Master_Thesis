```python
from pathlib import Path
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.metrics import accuracy_score, recall_score, f1_score

MISLABELS_CSV = Path("/home/student/lisa_ma/results/glare/22-09-26_12:48_ep-4_single-gpu/mislabels_list_20260922_124805.csv")
RC_XLSX       = Path("/home/student/lisa_ma/evaluation/gt_mislabels_complete.xlsx")
MAX_SCORE     = 4

scores_df = pd.read_csv(MISLABELS_CSV, dtype=str, keep_default_na=False)
scores_df["glare_score"] = pd.to_numeric(scores_df["glare score"], errors="coerce")
scores_df = scores_df.dropna(subset=["glare_score"]).reset_index(drop=True)
```

```python
# To get the mislabeled / clean label
df_rc     = pd.read_excel(RC_XLSX)                           # kein sheet_name nötig
positives = set(df_rc["sample_id"].astype(str))             # sample_id statt id, kein in_training_data Filter
scores_df["mislabeled"] = scores_df["id"].apply(lambda x: 1 if x in positives else 0)
```

```python
metric = "Recall"

metrics = []

for threshold in range(MAX_SCORE + 1):
    fraction = (scores_df["glare_score"] <= threshold).sum() / len(scores_df)

    y_true = scores_df["mislabeled"]
    y_pred = scores_df["glare_score"].apply(lambda s: 1 if s <= threshold else 0)

    accuracy = accuracy_score(y_true, y_pred)
    recall   = recall_score(y_true, y_pred, zero_division=0)
    f1       = f1_score(y_true, y_pred, zero_division=0)

    metrics.append({
        "Fraction": fraction,
        "Accuracy": accuracy,
        "Recall":   recall,
        "F1":       f1
    })

metrics_df = pd.DataFrame(metrics)

metrics_melted = metrics_df.melt(
    id_vars=["Fraction"],
    value_vars=["Accuracy", "Recall", "F1"],
    var_name="Metric",
    value_name="Value"
)

plt.rcParams.update({
    "font.family": "serif",
    "font.size": 12
})

df_filtered = metrics_melted[metrics_melted["Metric"] == metric]

plt.figure(figsize=(8, 6))

plt.ylim(0.0, 1.1)

plt.plot(df_filtered["Fraction"], df_filtered["Value"], color="b", linewidth=1)

plt.xlabel("Fraction of Training Data Checked")
plt.ylabel(metric)

plt.legend(frameon=True, loc="lower right")

plt.show()
```

```python
from sklearn.metrics import roc_auc_score, average_precision_score, precision_score

# ── Dataset summary ───────────────────────────────────────────────────────────
n_total     = len(scores_df)
n_mislabels = scores_df["mislabeled"].sum()
base_rate   = scores_df["mislabeled"].mean()

# ── Per-threshold metrics ─────────────────────────────────────────────────────
y_true   = scores_df["mislabeled"].values
inverted = MAX_SCORE - scores_df["glare_score"].values  # high = suspicious

auroc = roc_auc_score(y_true, inverted)
ap    = average_precision_score(y_true, inverted)

rows = []
for threshold in range(MAX_SCORE + 1):
    y_pred    = (scores_df["glare_score"] <= threshold).astype(int).values
    fraction  = y_pred.sum() / n_total
    recall    = recall_score(y_true, y_pred, zero_division=0)
    precision = precision_score(y_true, y_pred, zero_division=0)
    f1        = f1_score(y_true, y_pred, zero_division=0)
    rows.append({"threshold": threshold, "fraction": fraction,
                 "recall": recall, "precision": precision, "f1": f1})

detail_df = pd.DataFrame(rows)
best      = detail_df.loc[detail_df["f1"].idxmax()]

# ── Print ─────────────────────────────────────────────────────────────────────
print("=" * 52)
print("  GLARE Run Summary — VerSe Sep 22, 2026")
print("=" * 52)
print(f"  Training samples:              {n_total:,}")
print(f"  Confirmed mislabels:           {n_mislabels}")
print(f"  Base mislabel rate:            {base_rate:.2%}")
print(f"  Training epochs:               {MAX_SCORE}")
print(f"  GLARE score range:             0 – {MAX_SCORE}")
print()
print("  Detection Performance")
print("  " + "-" * 38)
print(f"  AUROC:                         {auroc:.4f}")
print(f"  Average Precision:             {ap:.4f}")
print(f"  Best F1:                       {best.f1:.4f}  (score ≤ {int(best.threshold)})")
print(f"    Recall at best F1:           {best.recall:.2%}")
print(f"    Precision at best F1:        {best.precision:.2%}")
print(f"    Fraction checked:            {best.fraction:.2%}")
print("=" * 52)
print()
print("  Full threshold table:")
print(detail_df.to_string(index=False, float_format=lambda v: f"{v:.4f}"))
```

```python
# ── Rank-based Recall Curve (ep-4, Sep 22 run) ───────────────────────────────
# Sort all training samples by score ascending (lower = more suspicious).
# Recall at position k = (mislabels in top-k) / total_mislabels.

# Load glarex score from the same CSV
scores_df["glarex_score"] = pd.to_numeric(scores_df.get("glarex score", pd.Series(dtype=float)), errors="coerce")

# Fall back to xlsx if glarex score not in CSV
if scores_df["glarex_score"].isna().all():
    MISLABELS_XLSX = Path("/home/student/lisa_ma/results/glare/22-09-26_12:48_ep-4_single-gpu/mislabels_list_20260922_124805.xlsx")
    xlsx_ep4 = pd.read_excel(MISLABELS_XLSX, dtype=str)
    xlsx_ep4["glarex_score"] = pd.to_numeric(xlsx_ep4.get("glarex score", pd.Series(dtype=float)), errors="coerce")
    scores_df = scores_df.merge(xlsx_ep4[["id", "glarex_score"]], on="id", how="left", suffixes=("_old", ""))
    scores_df = scores_df.drop(columns=["glarex_score_old"], errors="ignore")

n_total     = len(scores_df)
n_mislabels = int(scores_df["mislabeled"].sum())

def recall_curve(df, score_col):
    sorted_df = df.dropna(subset=[score_col]).sort_values(score_col, ascending=True).reset_index(drop=True)
    n         = len(sorted_df)
    n_pos     = int(sorted_df["mislabeled"].sum())
    fracs     = (np.arange(1, n + 1)) / n
    recalls   = sorted_df["mislabeled"].cumsum().values / n_pos
    return fracs, recalls

fracs_glare,  rec_glare  = recall_curve(scores_df, "glare_score")
fracs_glarex, rec_glarex = recall_curve(scores_df, "glarex_score")

plt.rcParams.update({"font.family": "serif", "font.size": 12})

fig, ax = plt.subplots(figsize=(8, 6))
ax.plot(fracs_glare,  rec_glare,  color="#2255AA", linewidth=1.5, label="GLARE (ep-4)")
ax.plot(fracs_glarex, rec_glarex, color="#CC4422", linewidth=1.5, label="GLAREx (ep-4)")

ax.set_xlabel("Fraction of Training Data Checked", fontsize=12)
ax.set_ylabel("Recall", fontsize=12)
ax.set_xlim(0, 1)
ax.set_ylim(0, 1.05)
ax.legend(fontsize=11, frameon=True, loc="lower right")
ax.set_facecolor("white")
ax.grid(False)
plt.tight_layout()
plt.savefig("/home/student/lisa_ma/figures/recall_curve_ep4.jpg", dpi=300, bbox_inches="tight")
plt.show()

print("Fraction checked → Recall (GLARE | GLAREx)")
for frac in [0.01, 0.05, 0.10, 0.20, 0.50]:
    i_g  = min(int(np.searchsorted(fracs_glare,  frac, side="right")) - 1, len(fracs_glare)  - 1)
    i_gx = min(int(np.searchsorted(fracs_glarex, frac, side="right")) - 1, len(fracs_glarex) - 1)
    print(f"  {frac:.0%} → {rec_glare[i_g]:.1%}  |  {rec_glarex[i_gx]:.1%}")
```

```python
# ── Rank-based F1 Curve (ep-4, Sep 22 run) ───────────────────────────────────
# At each position k: precision = found/k, recall = found/n_pos, F1 = harmonic mean.

def f1_curve(df, score_col):
    sorted_df = df.dropna(subset=[score_col]).sort_values(score_col, ascending=True).reset_index(drop=True)
    n         = len(sorted_df)
    n_pos     = int(sorted_df["mislabeled"].sum())
    fracs     = (np.arange(1, n + 1)) / n
    cum_found = sorted_df["mislabeled"].cumsum().values
    k         = np.arange(1, n + 1)
    precision = cum_found / k
    recall    = cum_found / n_pos
    denom     = precision + recall
    f1        = np.where(denom > 0, 2 * precision * recall / denom, 0.0)
    return fracs, f1

fracs_glare_f1,  f1_glare  = f1_curve(scores_df, "glare_score")
fracs_glarex_f1, f1_glarex = f1_curve(scores_df, "glarex_score")

plt.rcParams.update({"font.family": "serif", "font.size": 12})

fig, ax = plt.subplots(figsize=(8, 6))
ax.plot(fracs_glare_f1,  f1_glare,  color="#2255AA", linewidth=1.5, label="GLARE (ep-4)")
ax.plot(fracs_glarex_f1, f1_glarex, color="#CC4422", linewidth=1.5, label="GLAREx (ep-4)")

ax.set_xlabel("Fraction of Training Data Checked", fontsize=12)
ax.set_ylabel("F1", fontsize=12)
ax.set_xlim(0, 1)
ax.set_ylim(0, 1.05)
ax.legend(fontsize=11, frameon=True, loc="upper right")
ax.set_facecolor("white")
ax.grid(False)
plt.tight_layout()
plt.savefig("/home/student/lisa_ma/figures/f1_curve_ep4.jpg", dpi=300, bbox_inches="tight")
plt.show()

print("Fraction checked → F1 (GLARE | GLAREx)")
for frac in [0.01, 0.05, 0.10, 0.20, 0.50]:
    i_g  = min(int(np.searchsorted(fracs_glare_f1,  frac, side="right")) - 1, len(fracs_glare_f1)  - 1)
    i_gx = min(int(np.searchsorted(fracs_glarex_f1, frac, side="right")) - 1, len(fracs_glarex_f1) - 1)
    print(f"  {frac:.0%} → {f1_glare[i_g]:.3f}  |  {f1_glarex[i_gx]:.3f}")
```
